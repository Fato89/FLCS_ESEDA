## Corrección de tipos y recodificación

### ¿Qué está mal?

Polars infiere los tipos columna por columna. Lo hace bien para variables genuinamente numéricas como `fexp` o `talla`, pero se equivoca con columnas que contienen **códigos numéricos**: números que representan categorías.

Un `Float64` donde debería haber una categoría no genera error: simplemente devuelve resultados sin sentido. `df["area"].mean()` da 1.38. Nadie lo advierte.

```python
# Ver los tipos actuales de las variables que nos interesan
df.select(["area", "region", "prov", "etnia", "f1_s1_2","estrato", "parr_pri", "grupo_edad_nin"]).schema
```

### Recodificación con `pl.when().then().otherwise()`

Mapeamos los códigos numéricos a etiquetas legibles usando los valores del diccionario de variables de la ENDI. Al mismo tiempo declaramos el tipo correcto: `pl.Categorical` para variables nominales, `pl.Enum` para variables con orden natural.

```python
df = df.with_columns([

    # area: nominal, sin orden entre categorías
    pl.when(pl.col("area") == 1).then(pl.lit("Urbano"))
      .when(pl.col("area") == 2).then(pl.lit("Rural"))
      .otherwise(pl.lit(""))
      .cast(pl.Categorical)
      .alias("area"),

    # region: nominal, sin orden geográfico implícito
    pl.when(pl.col("region") == 1).then(pl.lit("Sierra"))
      .when(pl.col("region") == 2).then(pl.lit("Costa"))
      .when(pl.col("region") == 3).then(pl.lit("Amazonia"))
      .when(pl.col("region") == 4).then(pl.lit("Insular"))
      .otherwise(pl.lit(None))
      .cast(pl.Categorical)
      .alias("region"),

    # f1_s1_2 (sexo): nominal
    pl.when(pl.col("f1_s1_2") == 1).then(pl.lit("Hombre"))
      .when(pl.col("f1_s1_2") == 2).then(pl.lit("Mujer"))
      .otherwise(pl.lit(None))
      .cast(pl.Categorical)
      .alias("f1_s1_2"),

    # etnia: nominal
    pl.when(pl.col("etnia") == 1).then(pl.lit("Indígena"))
      .when(pl.col("etnia") == 2).then(pl.lit("Afroecuatoriana/o"))
      .when(pl.col("etnia") == 3).then(pl.lit("Montubia/o"))
      .when(pl.col("etnia") == 4).then(pl.lit("Mestiza/o"))
      .when(pl.col("etnia") == 5).then(pl.lit("Blanca/o"))
      .when(pl.col("etnia") == 6).then(pl.lit("Otra"))
      .otherwise(pl.lit(None))
      .cast(pl.Categorical)
      .alias("etnia"),

    # prov: código string
    pl.col("prov").cast(pl.Int8).cast(pl.String).str.pad_start(2,"0"),

    # estrato y parr_pri: categóricas de diseño muestral
    pl.col("parr_pri").cast(pl.Int8),

])

```

### Variables ordinales con `pl.Enum`

`grupo_edad_nin` tiene un orden natural: los grupos de edad van de menor a mayor. Con `pl.Enum` declaramos ese orden explícitamente. Polars lo usará al ordenar y en gráficos sin que tengamos que especificarlo cada vez.

```python
orden_grupos = [
    "0-5 meses",
    "6-11 meses",
    "12-23 meses",
    "24-35 meses",
    "36-47 meses",
    "48-59 meses",
]

df.with_columns(
    pl.when(pl.col("grupo_edad_nin") == 1).then(pl.lit(orden_grupos[0]))
      .when(pl.col("grupo_edad_nin") == 5).then(pl.lit(orden_grupos[4]))
      .when(pl.col("grupo_edad_nin") == 3).then(pl.lit(orden_grupos[2]))
      .when(pl.col("grupo_edad_nin") == 4).then(pl.lit(orden_grupos[3]))
      .when(pl.col("grupo_edad_nin") == 2).then(pl.lit(orden_grupos[1]))
      .when(pl.col("grupo_edad_nin") == 6).then(pl.lit(orden_grupos[5]))
      .otherwise(pl.lit(None))
      .cast(pl.Enum(orden_grupos))
      .alias("grupo_edad_nin")
)
# ["grupo_edad_nin"].unique()
```

##  Calcular edad en días desde las fechas crudas

### El problema

La variable `edaddias` ya viene calculada en la tabla. Sin embargo, vamos a recalcularla desde las columnas crudas por dos razones:

1. Para entender exactamente cómo se construye (el peso del denominador en la fórmula del indicador depende de esto)
2. Para aprender a trabajar con fechas en Polars, que es necesario en el indicador de vacunación de rotavirus

La lógica: edad en días = fecha de medición - fecha de nacimiento.

### Construir fechas desde columnas separadas

La ENDI guarda día, mes y año en columnas separadas. Hay que combinarlas en una fecha.

```python
# Revisar los tipos de las columnas de fecha antes de operar
print(df_nin.select(["nac_dia", "nac_mes", "nac_anio",
                     "med_dia", "med_mes", "med_anio"]).schema)
```

```python
# Construir las fechas
# pl.date(year, month, day) crea una columna de tipo Date a partir de componentes enteros
df_nin = df_nin.with_columns([
    pl.date(
        pl.col("nac_anio").cast(pl.Int32),
        pl.col("nac_mes").cast(pl.Int32),
        pl.col("nac_dia").cast(pl.Int32),
    ).alias("fecha_nac"),

    pl.date(
        pl.col("med_anio").cast(pl.Int32),
        pl.col("med_mes").cast(pl.Int32),
        pl.col("med_dia").cast(pl.Int32),
    ).alias("fecha_med"),
])

print(df_nin.select(["fecha_nac", "fecha_med"]).head(5))
```

### 4.3 Calcular edad en días

La diferencia entre dos fechas en Polars devuelve una duración. Con `.dt.total_days()` la convertimos a un entero.

```python
df_nin = df_nin.with_columns(
    (pl.col("fecha_med") - pl.col("fecha_nac"))
    .dt.total_days()
    .alias("edad_dias_calc")
)

# Comparar con la variable precalculada del INEC
print(
    df_nin.select([
        "edaddias",
        "edad_dias_calc",
        (pl.col("edaddias") - pl.col("edad_dias_calc")).abs().alias("diferencia")
    ]).describe()
)
```

```python
# ¿Hay casos donde difieren?
distintos = df_nin.filter(
    (pl.col("edaddias") - pl.col("edad_dias_calc")).abs() > 1
)
print(f"Casos con diferencia > 1 día: {distintos.shape[0]}")
```

##  Joins

### Tipos de join

| Tipo | ¿Qué conserva? | Cuándo usarlo |
|---|---|---|
| `left` | Todas las filas de la tabla izquierda | Agregar columnas de otra tabla sin perder casos |
| `right` | Todas las filas de la tabla derecha | Menos común; equivale a un `left` con las tablas invertidas |
| `inner` | Solo filas con coincidencia en ambas tablas | Cuando se quiere el subconjunto que existe en las dos |
| `full` | Todas las filas de ambas tablas | Cuando no se quiere perder ningún caso de ninguna tabla |
| `semi` | Filas de la izquierda que tienen coincidencia, sin agregar columnas | Filtrar sin expandir |
| `anti` | Filas de la izquierda que **no** tienen coincidencia | Encontrar casos que faltan en la otra tabla |

### Left join: personas con hogar

El caso más común en la ENDI: queremos agregar variables del hogar a la tabla de personas. Cada persona pertenece a un hogar (`id_hogar`), y queremos conservar todas las personas aunque algún hogar no tenga datos.

```python
# Cargar f1_hogar
ruta_hogar = "../../Materiales/Insumos/ENDI/BDD_ENDI_R2_rds/BDD_ENDI_R2_f1_hogar.rds"

df_hogar = pl.from_pandas(
    pyreadr.read_r(ruta_hogar)[None]
)

print(f"Hogares: {df_hogar.shape[0]:,} filas  |  {df_hogar.shape[1]} columnas")
```

```python
# Seleccionar solo las columnas de hogar que nos interesan
# f1_s3_2: tipo de vivienda
# f1_s7_1: fuente de agua para beber
df_hogar_min = df_hogar.select(["id_hogar", "f1_s3_2", "f1_s7_1"])

df_hogar_min = df_hogar_min.rename({
    "f1_s3_2":  "tipo_vivienda",
    "f1_s7_1": "fuente_agua",
})

print(df_hogar_min.head(3))
```

```python
# Left join: agregar variables del hogar a los niños
df_nin_hogar = df_nin.join(df_hogar_min,on="id_hogar",how="left")

print(f"Filas antes del join: {df_nin.shape[0]:,}")
print(f"Filas después:        {df_nin_hogar.shape[0]:,}")
print(f"\nNuevas columnas: {[c for c in df_nin_hogar.columns if c not in df_nin.columns]}")
```

```python
# Verificar: ¿algún niño quedó sin datos de hogar?
sin_hogar = df_nin_hogar.filter(pl.col("tipo_vivienda").is_null()).shape[0]
print(f"Niños sin datos de hogar después del join: {sin_hogar}")
```

### Inner join vs left join - la diferencia

Para ver la diferencia concretamente, comparamos el número de filas que devuelve cada tipo.

```python
n_left  = df_nin.join(df_hogar_min, on="id_hogar", how="left").shape[0]
n_inner = df_nin.join(df_hogar_min, on="id_hogar", how="inner").shape[0]
n_anti  = df_nin.join(df_hogar_min, on="id_hogar", how="anti").shape[0]

print(f"left join:  {n_left:>7,} filas  (conserva todos los niños)")
print(f"inner join: {n_inner:>7,} filas  (solo niños con hogar coincidente)")
print(f"anti join:  {n_anti:>7,} filas  (niños sin hogar coincidente)")
print(f"\nSuma inner + anti = {n_inner + n_anti:,}  (debe coincidir con el total de niños)")
```
