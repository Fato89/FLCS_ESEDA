```python
import polars as pl
```

```python
x= pl.DataFrame({
    "a":["a","b","c"],
    "b":["t","u","v"],
    "c":["1","2","3"],
        })

y= pl.DataFrame({
    "a":["a","b","d"],
    "b":["t","u","w"],
    "c":["4","5","6"],
        })

pl.concat([x,y], how='vertical')
```

```python
x= pl.DataFrame({
    "a":["a","b","c"],
    "b":["t","u","v"],
    "c":["1","2","3"],
        })

y= pl.DataFrame({
    "d":["a","b","d"],
    "e":["t","u","w"],
    "f":["4","5","6"],
        })

pl.concat([x,y], how='horizontal')
```

```python
x= pl.DataFrame({
    "a":["a","b","c"],
    "b":["t","u","v"],
    "c":["1","2","3"],
        })

y= pl.DataFrame({
    "a":["a","b","d"],
    "b":["t","u","w"],
    "d":["3","2","1"],
        })
```

```python
x.join(y,on="a",suffix='_r',how="left")
```

```python
x.join(y,on="a",suffix='_r',how="inner")
```

```python
x.join(y,on="a",suffix='_r',how="full").sort("a")
```

```python
x.join(y,on="a",suffix='_r',how="semi")
```

```python
x.join(y,on="a",suffix='_r',how="anti")
```
